```python
# condicionales.py
import pandas as pd

# 1. Leer la hoja Datos_Climaticos (encabezados en la fila 4 -> header=3)
df = pd.read_excel(
    "monitoreo_climatico_ambiental.xlsx",
    sheet_name="Datos_Climaticos",
    header=3,
    nrows=60
)

# Seleccionamos una fila puntual sin usar bucles (ej. índice 19 corresponde a REG-020)
muestra = df.iloc[19]

print("=== EVALUACIÓN INDIVIDUAL POR CONDICIONALES ===")
print("Registro:", muestra["ID_Registro"])
print("Fecha:", str(muestra["Fecha"])[:10])
print("Estación:", muestra["Estación"])
print("-" * 50)

# 1. Condicional múltiple (if / elif / else): Calidad del aire por PM2.5 (Res. 2254 de 2017)
pm25 = muestra["PM2.5 (µg/m³)"]

if pm25 <= 15.0:
    estado_aire = "Buena (Cumple guía OMS y norma nacional)"
elif pm25 <= 37.0:
    estado_aire = "Moderada (Cumple límite diario nacional <= 37 µg/m³)"
else:
    estado_aire = "Alerta Ambiental (Excede límite diario nacional)"

print("Calidad del aire (PM2.5 =", pm25, "µg/m³):", estado_aire)

# 2. Condicional compuesto con operador lógico (and): Precipitación y Humedad Relativa
lluvia = muestra["Precipitación (mm)"]
humedad = muestra["Humedad_Rel (%)"]

if lluvia >= 25.0 and humedad >= 85.0:
    alerta_hidrica = "Alerta Amarilla: Lluvia fuerte con alta saturación atmosférica"
elif lluvia > 0.0:
    alerta_hidrica = "Normal: Evento de lluvia ligera a moderada"
else:
    alerta_hidrica = "Tiempo seco: Sin precipitación registrada"

print("Diagnóstico hídrico (Lluvia =", lluvia, "mm | HR =", humedad, "%):", alerta_hidrica)

# 3. Condicional anidado: Evaluación térmica y riesgo de heladas en sabana
temp_min = muestra["Temp_Min (°C)"]
temp_max = muestra["Temp_Max (°C)"]
temp_media = round((temp_min + temp_max) / 2, 1)

if temp_min < 4.0:
    riesgo_termico = "Alerta de Helada Radiativa (Temp. mínima < 4.0 °C)"
else:
    if temp_media >= 12.0 and temp_media <= 16.0:
        riesgo_termico = "Rango térmico típico de sabana (" + str(temp_media) + " °C)"
    else:
        riesgo_termico = "Variación fuera del promedio habitual (" + str(temp_media) + " °C)"

print("Comportamiento térmico:", riesgo_termico)
```

```python
# bucles.py
import pandas as pd

# 1. Cargar los 60 registros desde la hoja Datos_Climaticos (encabezados en la fila 4 -> header=3)
df = pd.read_excel(
    "monitoreo_climatico_ambiental.xlsx",
    sheet_name="Datos_Climaticos",
    header=3,
    nrows=60
)

# Convertimos la tabla en una lista de diccionarios para recorrerla con bucles puros
registros = df.to_dict(orient="records")
total_registros = len(registros)

print("1. BUCLE FOR: ESTADÍSTICAS ACUMULADAS")

lluvia_acumulada = 0.0
suma_pm25 = 0.0
alertas_pm25 = 0

for fila in registros:
    lluvia_acumulada = lluvia_acumulada + fila["Precipitación (mm)"]
    suma_pm25 = suma_pm25 + fila["PM2.5 (µg/m³)"]
    
    # Contar registros que superan el umbral diario de 37.0 µg/m³ (Res. 2254 de 2017)
    if fila["PM2.5 (µg/m³)"] > 37.0:
        alertas_pm25 = alertas_pm25 + 1

promedio_pm25 = suma_pm25 / total_registros

print("Total de mediciones procesadas:", total_registros)
print("Precipitación total acumulada:", round(lluvia_acumulada, 1), "mm")
print("Concentración media de PM2.5:", round(promedio_pm25, 2), "µg/m³")
print("Mediciones en Alerta Ambiental (> 37 µg/m³):", alertas_pm25)

print("\n 2. BUCLE WHILE: BÚSQUEDA DEL PRIMER EVENTO DE LLUVIA FUERTE")

# Buscar cronológicamente el primer registro con precipitación >= 25.0 mm (umbral IDEAM)
indice = 0
evento_encontrado = False

while indice < total_registros and not evento_encontrado:
    dato = registros[indice]
    if dato["Precipitación (mm)"] >= 25.0:
        print("Primer evento crítico hallado en:", dato["ID_Registro"])
        print("Fecha:", str(dato["Fecha"])[:10])
        print("Estación:", dato["Estación"])
        print("Precipitación registrada:", dato["Precipitación (mm)"], "mm")
        evento_encontrado = True
    indice = indice + 1
```

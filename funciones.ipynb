```python
import math
import pandas as pd


def calcular_presion_saturacion(temperatura_c):
    """Función exponencial f(T): Presión de vapor de saturación en kPa (Tetens)."""
    exponente = (17.27 * temperatura_c) / (temperatura_c + 237.3)
    e_s = 0.6108 * math.exp(exponente)
    return round(e_s, 4)


def calcular_punto_rocio(temperatura_c, humedad_relativa):
    """Función logarítmica f(T, HR): Punto de rocío en °C (Magnus-Tetens)."""
    gamma = math.log(humedad_relativa / 100.0) + (
        (17.27 * temperatura_c) / (temperatura_c + 237.3)
    )
    t_rocio = (237.3 * gamma) / (17.27 - gamma)
    return round(t_rocio, 2)


def calcular_evapotranspiracion(t_min, t_max, radiacion_wm2):
    """Función con raíz cuadrada f(Tmin, Tmax, Rad): Evapotranspiración ETo en mm/día."""
    t_media = (t_min + t_max) / 2.0
    rad_mm_dia = (radiacion_wm2 * 0.0864) / 2.45
    eto = 0.0023 * rad_mm_dia * (t_media + 17.8) * math.sqrt(t_max - t_min)
    return round(eto, 2)


def calcular_presion_barometrica(altitud_msnm):
    """Función potencial f(h): Presión atmosférica en kPa según la altitud."""
    base = (293.0 - 0.0065 * altitud_msnm) / 293.0
    presion = 101.3 * math.pow(base, 5.26)
    return round(presion, 2)


# === LECTURA DEL EXCEL Y EVALUACIÓN MATEMÁTICA ===
df = pd.read_excel(
    "monitoreo_climatico_ambiental.xlsx",
    sheet_name="Datos_Climaticos",
    header=3,
    nrows=60
)

# Tomamos el primer registro (REG-001: EST-01 Facatativá Centro)
fila = df.iloc[0]
t_min = fila["Temp_Min (°C)"]
t_max = fila["Temp_Max (°C)"]
t_med = (t_min + t_max) / 2.0
hr = fila["Humedad_Rel (%)"]
rad = fila["Rad_Solar (W/m²)"]
altitud = fila["Altitud (msnm)"]

print("=== EVALUACIÓN DE FUNCIONES MATEMÁTICAS CLIMÁTICAS ===")
print(f"Registro: {fila['ID_Registro']} | Estación: {fila['Estación']} ({altitud} msnm)")
print(f"Variables de entrada -> T_med: {t_med} °C | HR: {hr}% | Rad: {rad} W/m²")
print("-" * 58)

print("1. Presión de saturación e_s(T):", calcular_presion_saturacion(t_med), "kPa")
print("2. Punto de rocío Td(T, HR):", calcular_punto_rocio(t_med, hr), "°C")
print("3. Evapotranspiración potencial ETo:", calcular_evapotranspiracion(t_min, t_max, rad), "mm/día")
print("4. Presión atmosférica P(h):", calcular_presion_barometrica(altitud), "kPa")

# Muestra la tabla del Excel debajo de los resultados
display(df)
```
